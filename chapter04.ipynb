## 딕셔너리 (Dictionaray)


- {"키": "값"}
- "키" 중복 불가
- "값" 중복 가능
- 추가하는 순서를 유지

```python
{"a":1}
{"a":1,"b":1}
# print({"a":1,"b":2})
print({"a":1,"b":1}) 
```

```python
user_info = {
    "name":"Kimjun",
    "age":28
}

print(user_info)
print(user_info["age"]) #키로 값을 가져오기
user_info["age"] = 30
print(user_info["age"])
user_info["role"] = "Tutor"
print(user_info)

# user_info[0]
```

### 삭제

```python
user_info = {
    "name":"Kimjun",
    "age":28,
    "role": "admin"
}
print(user_info)
# del 메서드 사용
del user_info["role"]

print(user_info)

# pop 사용 : 꺼내면서 삭제
age = user_info.pop("age")
print(age,user_info)

```

### 딕셔너리 주요 메서드

- keys(), values(), items()

```python
menu = {"coffee" : 4000, "tea": 4500}

# Key 목록 족회
print(menu.keys())

# value 목록 조회
print(menu.values())

# itmes 조회 (키, 값) 튜플 형태로 리텅
print(menu.items()) 

# 딕셔너리는 인덱서 방식으로 값을 참조 불가 > List 로 변환
print(list(menu.keys())[1])

print(sum(menu.values()))
```

```python
print("coffee" in menu) #True
print(4000 in menu) #False

print(4000 in menu.values()) #True
```

```python
# menu["cake"]
price =  menu.get("cake", 5000)
print(price)

# default 값이 없는 경우
print(menu.get("cake"))

print(menu.get("coffee",0))
print(menu)
```

### 집합 (Set)


수학 집합 : 중복 없고, 순서 없는 자료형

- 중괄호 사용
- set()함수를 생성


```python
grade = set([1, 2, 2, 3, 3, 3])

print (grade)
```

```python
print (type({})) # {} 빈 딕셔너리
print (type(set())) # set() 빈 집합
print ({1,2,2}) # 집합
```

```python
s = {1,2}

s.add(2)
s.add(5)

print(s)

s.remove(1)

print(s)

s.discard(20) # 없는 키 삭제할 때도 오류가 없음
print(s)
```

```python
set_a = {1,2,3}
set_b = {3,4,5}

print(set_a & set_b) # 교집합
print(set_a | set_b) # 합집합
print(set_a - set_b) # 차집합
print(3 in set_a) # 포함 여부


```

### 중복 제거

```python
raw_emails = ["a@sparta.com","b@sparta.com","a@sparta.com"]

# 집합으로 중복 제거
unique_sets =  set(raw_emails)


print(list(unique_sets))

print("# 중복제거 및 순서 유지")
sort_emails = list(dict.fromkeys(raw_emails))
print(sort_emails)

# sorted

print(sorted(set(["c","b","a","f","a"])))



```
