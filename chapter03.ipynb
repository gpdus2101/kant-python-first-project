````python
# 리스트 : 가변 데이터 저장
shopping_list = ["apple","banana","cherry"]

print(shopping_list)
print(shopping_list[0])

shopping_list[0] = "kiwi"
print(shopping_list[0])
print(shopping_list)
````

---
### 리스트 메서드

````python
# append
fruits = ["apple"]
fruits.append("banana")
print(fruits)
# insert
fruits.insert(0,"kiwi")
print(fruits)
# remove
fruits.remove("apple")
print(fruits)

# fruits.remove("grape")
````

### 슬라이싱

````python
numbers = [0,1,2,3,4,5]

# 1번 부터 4번 미만까지
numbers[1:4]
# 2칸 간격으로 추출
numbers[::2]
# 역순으로 추출
numbers[::-1]
# 앞에서 3개만 추출
numbers[:3]
# 3분부터 끝까지
numbers[3:]
````

````python
print((10,20,30,40)[1:3])
````
